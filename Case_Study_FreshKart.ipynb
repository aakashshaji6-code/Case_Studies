```python
#Task 6
```

```python
# Check column names
print(internet.columns)
print(population.columns)
print(gdp.columns)
print(literacy.columns)


gdp.rename(columns={
    'Country': 'country',
    'GDP': 'gdp'
}, inplace=True)
literacy.rename(columns={
    'name': 'country',
    'Literacy Rate': 'literacy_rate'
}, inplace=True)
```

```python
#Task 7
```

```python
gdp.info()

gdp['gdp'] = pd.to_numeric(
    gdp['gdp'], errors='coerce')
```

```python
literacy.info()
literacy['literacy_rate'] = pd.to_numeric(
    literacy['literacy_rate'], errors='coerce')
```

```python
internet.info()
internet['internet_users'] = pd.to_numeric(
    internet['internet_users'], errors='coerce')
```

```python
population.info()
population['population'] = pd.to_numeric(
    population['population'], errors='coerce'
)
```

```python
#Task 8
```

```python
#Why use outer join?
#Because we want to combine all four datasets without losing any country records.
#If a country is missing from one dataset, its corresponding values will appear as NaN.

finaldf = pd.merge(population, gdp, on='country', how='outer')

finaldf = pd.merge(finaldf, internet, on='country', how='outer')

finaldf = pd.merge(finaldf, literacy, on='country', how='outer')

finaldf.head()
```

```python
#Task 9
```

```python
finaldf['internet_penetration'] = ( finaldf['internet_users'] / finaldf['population']) * 100

finaldf['internet_penetration'] = finaldf[ 'internet_penetration'].round(2)
finaldf.head()
```

```python
finaldf['gdp_per_capita'] = (finaldf['gdp'] / finaldf['population'])

finaldf['gdp_per_capita'] = finaldf['gdp_per_capita'].round(2)
finaldf.head()
```

```python
#Task 10
```

```python
import sqlite3

conn = sqlite3.connect('/content/country.db')

finaldf.to_sql(
    'country_statistics',
    conn,
    if_exists='replace',
    index=False
)

print("Data successfully stored in database")
```
