```python
import numpy as np
import pandas as pd

np.random.seed(42)
n = 200

df = pd.DataFrame({
    "order_id": range(1001, 1001 + n),
    "order_date": pd.date_range("2024-01-01", periods=n, freq="D"),
    "customer_name": np.random.choice(
        ["Anita", "Ravi", "Meera", "Arjun", "Sneha", "Karthik", "Divya", "Suresh"], n),
    "city": np.random.choice(
        ["Kochi", "Chennai", "Bengaluru", "Hyderabad", "Mumbai"], n),
    "category": np.random.choice(
        ["Electronics", "Clothing", "Grocery", "Books", "Furniture"], n),
    "quantity": np.random.randint(1, 10, n),
    "unit_price": np.random.randint(100, 5000, n),
    "payment_mode": np.random.choice(
        ["UPI", "Card", "Cash", "NetBanking"], n)
})

df.to_csv("sales_data.csv", index=False)
print("File created with", len(df), "rows")
```

```python
def calculate_total(quantity, unit_price, discount_percent=0):

    total = quantity * unit_price

    discount = total * discount_percent / 100

    final_amount = total - discount

    return final_amount
```

```python
amounts = [1200, 8500, 4999, 5000, 300, 15000, 7200, 950]

high = 0
low = 0

for amount in amounts:
    if amount >= 5000:
        high += 1
    else:
        low += 1

print(f"High amounts: {high}")
print(f"Low amounts: {low}")
```

```python
import numpy as np
import pandas as pd
prices = np.array([250, 1200, 450, 3000, 780, 1500, 90, 2200, 640, 4100])
```

```python
print("Mean:", np.mean(prices))

print("Median:", np.median(prices))

print("Minimum:", np.min(prices))

print("Maximum:", np.max(prices))

print("Standard Deviation:", np.std(prices))
```

```python
high_prices = prices[prices > 1000]

print("Prices greater than 1000:", high_prices)

print("Count:", len(high_prices))
```

```python
new_prices = prices.reshape(2, 5)

print("Reshaped Array:")
print(new_prices)

print("Row Sum:", np.sum(new_prices, axis=1))

print("Column Sum:", np.sum(new_prices, axis=0))
```

```python
labels = np.where(prices > 2000, "Expensive", "Affordable")

print(labels)
```

```python
import pandas as pd
sales = pd.read_csv("sales_data.csv", parse_dates=["order_date"])
```

```python
print(sales.head())

print(sales.tail(3))

print(sales.shape)

print(sales.columns)

print(sales.dtypes)

print(sales.describe())

print(sales.isnull().sum())
```

```python
sales["total_amount"] = sales["quantity"] * sales["unit_price"]
sales["month"] = sales["order_date"].dt.month_name()

display(sales.head())
```

```python
result = sales[(sales["city"] == "Kochi") & (sales["category"] == "Electronics")]

print(result)
```

```python
result = sales[(sales["total_amount"] > 20000) | (sales["payment_mode"] == "Cash")]

print(result)

print("Number of rows:", len(result))
```

```python
print(sales.groupby("category")["total_amount"].sum())
```

```python
print(sales.groupby("city")["total_amount"].mean())
```

```python
print(sales.groupby("payment_mode")["order_id"].count())
```

```python
customer_sales = sales.groupby("customer_name")["total_amount"].sum()

print(customer_sales.sort_values(ascending=False).head(3))
```

```python
highest_orders = sales.sort_values("total_amount", ascending=False)

print(highest_orders.head(5))
```

```python
pivot = sales.pivot_table(
    values="total_amount",
    index="city",
    columns="category",
    aggfunc="sum"
)

print(pivot)
```

```python
import matplotlib.pyplot as plt

category_sales = sales.groupby("category")["total_amount"].sum()

plt.bar(category_sales.index, category_sales.values,
        color=["red", "blue", "green", "orange", "purple"])

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
monthly_sales = sales.groupby("month")["total_amount"].sum()

month_order = ["January", "February", "March", "April",
               "May", "June", "July"]

monthly_sales = monthly_sales.reindex(month_order)

plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Total Sales")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.grid()

plt.show()
```

```python
import seaborn as sns

sns.boxplot(x="category", y="total_amount", data=sales)

plt.title("Total Amount by Category")
plt.xlabel("Category")
plt.ylabel("Total Amount")

plt.show()
```

```python
sns.countplot(x="payment_mode", data=sales)

plt.title("Orders by Payment Mode")
plt.xlabel("Payment Mode")
plt.ylabel("Number of Orders")

plt.show()
```

```python
import sqlite3

conn = sqlite3.connect("sales.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY,
    order_date TEXT,
    customer_name TEXT,
    city TEXT,
    category TEXT,
    quantity INTEGER,
    unit_price INTEGER,
    payment_mode TEXT
)
""")

conn.commit()
```

```python
sales["order_date"] = sales["order_date"].astype(str)

sales.to_sql("orders", conn, if_exists="replace", index=False)

result = pd.read_sql_query("SELECT COUNT(*) FROM orders", conn)

print(result)
```

```python
query = "SELECT * FROM orders WHERE city = 'Mumbai'"

result = pd.read_sql_query(query, conn)

print(result)
```

```python
query = """
SELECT category, SUM(quantity) AS total_quantity
FROM orders
GROUP BY category
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
query = """
SELECT city, AVG(unit_price) AS average_price
FROM orders
GROUP BY city
HAVING AVG(unit_price) > 2500
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
query = """
SELECT *, quantity * unit_price AS total_amount
FROM orders
ORDER BY total_amount DESC
LIMIT 5
"""

result = pd.read_sql_query(query, conn)

print(result)
```
