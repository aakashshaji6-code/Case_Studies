```python
print(df[["Revenue (Millions)", "Metascore"]].isna().sum())
```

```python
revenue_mean= df["Revenue (Millions)"].mean()

metascore_mean= df["Metascore"].mean()

print("Revenue Mean:", revenue_mean)
print("Metascore Mean:", metascore_mean)
```

```python
df["Revenue (Millions)"] = df["Revenue (Millions)"].fillna(revenue_mean)
df["Metascore"] = df["Metascore"].fillna(metascore_mean)
```

```python
print(df[["Revenue (Millions)", "Metascore"]].isna().sum())
```

```python
# Q16 - Create Profitability column

df['Profitability'] = df['Revenue (Millions)'] - df['Metascore']

df.head()

df.nlargest(10, 'Profitability')[['Title', 'Profitability']]
```

```python
def category(rating):
    if rating < 6:
        return 'Poor'
    elif 6 <= rating < 7.5:
        return 'Average'
    else:
        return 'Excellent'
df['Rating_Category'] = df['Rating'].apply(category)
df[['Title', 'Rating', 'Rating_Category']]
```

```python
def score(row):
    return row['Rating'] * 10 + row['Revenue (Millions)']
df['High_Score'] = df.apply(score, axis=1)
df[['Title', 'Rating', 'Revenue (Millions)', 'High_Score']]
```
