```python
# 1 display each order
import sqlite3
import pandas as pd

# Re-open connection if closed
conn = sqlite3.connect("QuickCart_DB.db")
conn.execute("PRAGMA foreign_keys = ON;")

query15 = """
SELECT o.order_id, o.order_date, c.customer_id, c.customer_name
FROM Orders o
INNER JOIN Customers c ON o.customer_id = c.customer_id;
"""

pd.read_sql_query(query15, conn)
```

```python
#2 number of orders placed
import sqlite3
import pandas as pd

# Ensure database connection is active
conn = sqlite3.connect("QuickCart_DB.db")

query16 = """
SELECT c.customer_id, c.customer_name, COUNT(o.order_id) AS total_orders
FROM Customers c
LEFT JOIN Orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name;
"""

pd.read_sql_query(query16, conn)
```

```python
# 3 calculate the total amount
import sqlite3
import pandas as pd

# Re-establish connection
conn = sqlite3.connect("QuickCart_DB.db")

query17 = """
SELECT c.customer_id, c.customer_name,
       COALESCE(SUM(oi.quantity * oi.selling_price), 0) AS total_spent
FROM Customers c
LEFT JOIN Orders o ON c.customer_id = o.customer_id
LEFT JOIN Order_Items oi ON o.order_id = oi.order_id
GROUP BY c.customer_id, c.customer_name;
"""

pd.read_sql_query(query17, conn)
```

```python
#4 identify the customer who spent the most
import sqlite3
import pandas as pd

#  connection is open
conn = sqlite3.connect("QuickCart_DB.db")

query18 = """
SELECT c.customer_id, c.customer_name,
       SUM(oi.quantity * oi.selling_price) AS total_spent
FROM Customers c
INNER JOIN Orders o ON c.customer_id = o.customer_id
INNER JOIN Order_Items oi ON o.order_id = oi.order_id
GROUP BY c.customer_id, c.customer_name
ORDER BY total_spent DESC
LIMIT 1;
"""

pd.read_sql_query(query18, conn)
```

```python
#5 find customers who never placed an order
import sqlite3
import pandas as pd

# Ensure connection is open
conn = sqlite3.connect("QuickCart_DB.db")

query19 = """
SELECT c.customer_id, c.customer_name
FROM Customers c
LEFT JOIN Orders o ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL;
"""

pd.read_sql_query(query19, conn)
```

```python
# 6 # Find customers who have placed more than two orders
import sqlite3
import pandas as pd

# Re-establish connection explicitly to the absolute database path
conn = sqlite3.connect("/content/QuickCart_DB.db")
conn.execute("PRAGMA foreign_keys = ON;")


query6 = """
SELECT c.customer_name, COUNT(o.order_id) AS total_orders
FROM Customers c
JOIN Orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name
HAVING COUNT(o.order_id) > 2;
"""

pd.read_sql_query(query6, conn)
```

```python
#Display orders with their corresponding payment status.
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query1 = """
SELECT orders.order_id, orders.order_date, orders.order_status,
payments.payment_method, payments.payment_status,payments.amount
FROM Orders
INNER JOIN Payments
ON orders.order_id = payments.order_id;
"""
pd.read_sql_query(query1, conn)
```

```python
#Find all unpaid orders
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query2 = """
SELECT orders.order_id, orders.order_status, payments.payment_status, payments.amount
FROM Orders  INNER JOIN Payments  ON orders.order_id = payments.order_id
WHERE payments.payment_status = 'Unpaid';
"""

pd.read_sql_query(query2, conn)
```

```python
#Find all cancelled orders.
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query3 = """ SELECT * FROM Orders WHERE order_status = 'Cancelled'; """

pd.read_sql_query(query3, conn)
```

```python
#Calculate the total amount received through each payment method.
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query4 = """
SELECT payment_method, SUM(amount) AS Total_Amount_Received
FROM Payments WHERE payment_status = 'Paid'
GROUP BY payment_method;
"""

pd.read_sql_query(query4, conn)
```

```python
#Count successful and failed payments.
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query5 = """
SELECT payment_status, COUNT(payment_id) AS Total_Payments
FROM Payments GROUP BY payment_status;
"""

pd.read_sql_query(query5, conn)
```

```python
#Find orders where the order status is Delivered but the payment status is not Paid.
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query6 = """
SELECT orders.order_id, orders.order_status, payments.payment_status, payments.amount
FROM Orders  INNER JOIN Payments  ON orders.order_id = payments.order_id
WHERE orders.order_status = 'Delivered' AND payments.payment_status != 'Paid';
"""

pd.read_sql_query(query6, conn)
```

```python
#Identify the most commonly used payment method.
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query7 = """
SELECT payment_method, COUNT(payment_id) AS Total_Payments
FROM Payments GROUP BY payment_method
ORDER BY Total_Payments DESC
LIMIT 1;
"""

pd.read_sql_query(query7, conn)
```

```python
# Which product category should receive more attention from the marketing team?
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query1 = """ SELECT p.category, SUM(oi.quantity * oi.selling_price) AS total_sales FROM Products p JOIN Order_Items oi
ON p.product_id = oi.product_id
GROUP BY p.category
ORDER BY total_sales ASC;
"""
pd.read_sql_query(query1, conn)
```

```python
# Which products need restocking?
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query2 = """ SELECT product_id, product_name, stock_quantity FROM Products
WHERE stock_quantity < 10
ORDER BY stock_quantity ASC;
"""
pd.read_sql_query(query2, conn)
```

```python
# Find products that have never been ordered.
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query5 = """ SELECT p.product_name FROM Products p LEFT JOIN Order_Items oi
ON p.product_id = oi.product_id
WHERE oi.order_id IS NULL;
"""
pd.read_sql_query(query5, conn)
```

```python
# Find orders containing more than one product line.
import sqlite3
import pandas as pd
# Ensure we have a valid database connection

conn = sqlite3.connect("QuickCart_DB.db")
query6 = """ SELECT order_id, COUNT(*) AS total_product_lines
FROM Order_Items
GROUP BY order_id
HAVING COUNT(*) > 1;
"""
pd.read_sql_query(query6, conn)
```
